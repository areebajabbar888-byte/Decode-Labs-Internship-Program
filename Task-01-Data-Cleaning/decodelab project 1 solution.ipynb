```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\SSC\Downloads\Dataset for Data Analytics - Sheet1.csv")
df.head()
```

Understand the Dataset

```python
#check nummber of rows and columns
print("Dataset Shape",df.shape)
#check columns name
print("Column names:")
print(df.columns.tolist())
#check datatype and missing values
df.info()
#Display dataset
df.head()
```

Indentifying Missing values

```python
#count missing values in each column
print(df.isnull().sum())
#calculate missing value percentage
missing_percentage=(df.isnull().sum()/len(df))*100
print("\nMissing value percentage:")
print(missing_percentage)
```

Remove Duplicate records

```python
#check duplicated records
print("Total Duplicated:",df.duplicated().sum())
#Remove Duplicated records
df=df.drop_duplicates()
#verify duplicates
print("verify duplicates:",df.duplicated().sum())
```

Clean columns name and text

```python
#Remove extra spaces from columns name
df.columns=df.columns.str.strip()
#Remove extra spaces from text values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
#standardize text capitalization
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].apply(
        lambda x: x.title() if isinstance(x, str) else x
    )

df.head()
```

Correct date format

```python
#convert date column into datetime format
df["Date"]=pd.to_datetime(
    df["Date"],errors="coerce"
)
#verify date format
print(df["Date"].dtype)
df.head()
```

Correct Numerical datatypes

```python
#identify numeric columns
numeric_columns = [
    "Quantity",
    "UnitPrice",
    "ItemsInCart",
    "TotalPrice"
]

# Convert columns into numeric format
for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

# Verify data types
df.dtypes
```

Handle incorrect values

```python
# Replace invalid negative or zero quantities
df.loc[df["Quantity"] <= 0, "Quantity"] = np.nan

# Replace negative prices
df.loc[df["UnitPrice"] < 0, "UnitPrice"] = np.nan

df.loc[df["TotalPrice"] < 0, "TotalPrice"] = np.nan

# Check invalid values
print("Invalid Quantity:", df["Quantity"].isnull().sum())
print("Invalid Unit Price:", df["UnitPrice"].isnull().sum())
print("Invalid Total Price:", df["TotalPrice"].isnull().sum())
```

Handling Missing values

```python
# Fill missing numerical values with median
for col in numeric_columns:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())

# Fill missing text values
for col in df.select_dtypes(include="object").columns:

    if col not in ["OrderID", "CustomerID", "TrackingNumber"]:
        df[col] = df[col].fillna("Unknown")

# Check remaining missing values
df.isnull().sum()
```

Final verification

```python
# Check final dataset
print("Final Dataset Shape:", df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Records:")
print(df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)

df.head(10)
```

Export Cleaned Dataset

```python
# Save cleaned dataset
df.to_csv("cleaned_ecommerce_orders.csv", index=False)

print("Cleaned dataset exported successfully!")
```
